Import Libraries and Create Data

```python
import numpy as np
import matplotlib.pyplot as plt

subjects = np.array(["Maths", "Python", "Java", "DBMS", "AI"])

np.random.seed(10)
marks = np.random.randint(35, 100, (50, 5))

print(marks)
```

Inspect Array and Print Summary

```python
print("Number of Students:", marks.shape[0])
print("Number of Subjects:", marks.shape[1])
print("Shape:", marks.shape)
print("Dimensions:", marks.ndim)
print("Total Marks Entries:", marks.size)
print("Data Type:", marks.dtype)
print("Total Marks:", np.sum(marks))
print("Average Marks:", np.mean(marks))
```

Slicing and Indexing

```python
print("First 10 Students:")
print(marks[:10, :])

print("Last Subject Marks:")
print(marks[:, -1])

print("First 3 Subjects:")
print(marks[:, :3])

print("Subjects 2 to 4:")
print(marks[:, 1:4])
```

Boolean Masks

```python
below_40 = marks < 40
above_90 = marks > 90

print("Marks Below 40:")
print(marks[below_40])

print("Marks Above 90:")
print(marks[above_90])

print("Number of Marks Below 40:", np.sum(below_40))
print("Number of Marks Above 90:", np.sum(above_90))
```

Grade Bands

```python
grade_a = marks >= 80
grade_b = (marks >= 60) & (marks < 80)
grade_c = (marks >= 40) & (marks < 60)
grade_f = marks < 40

print("A Grade Marks:", np.sum(grade_a))
print("B Grade Marks:", np.sum(grade_b))
print("C Grade Marks:", np.sum(grade_c))
print("Below 40 Marks:", np.sum(grade_f))
```

Subject-Wise Toppers

```python
topper_marks = np.max(marks, axis=0)
topper_students = np.argmax(marks, axis=0)

print("Subject-Wise Toppers")

for i in range(5):
    print(subjects[i], "Top Marks:", topper_marks[i],
          "Student:", topper_students[i] + 1)
```

Subject-Wise Average

```python
subject_average = np.mean(marks, axis=0)

print("Subject-Wise Average")

for i in range(5):
    print(subjects[i], subject_average[i])
```

Pass Percentage

```python
pass_marks = marks >= 40
pass_count = np.sum(pass_marks, axis=0)
pass_percentage = (pass_count / 50) * 100

print("Subject-Wise Pass Percentage")

for i in range(5):
    print(subjects[i], pass_percentage[i], "%")
```

Student-Wise Result

```python
student_totals = np.sum(marks, axis=1)
student_average = np.mean(marks, axis=1)

print("Student-Wise Results")

for i in range(50):
    if np.all(marks[i] >= 40):
        result = "Pass"
    else:
        result = "Fail"

    print("Student", i + 1,
          "Total:", student_totals[i],
          "Average:", student_average[i],
          "Result:", result)
```

Find Class Topper and Failures

```python
class_topper = np.argmax(student_totals)
failed_students = np.where(np.any(marks < 40, axis=1))[0]

print("Class Topper: Student", class_topper + 1)
print("Topper Total Marks:", student_totals[class_topper])

print("Failed Students:")

for i in failed_students:
    print("Student", i + 1)

print("Total Failed Students:", len(failed_students))
print("Total Passed Students:", 50 - len(failed_students))
```

Subject-Wise Average Graph

```python
plt.figure(figsize=(8, 5))
plt.bar(subjects, subject_average)
plt.title("Subject-Wise Average Marks")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.ylim(0, 100)
plt.show()
```

Grade Distribution Graph

```python
grade_names = ["A Grade", "B Grade", "C Grade", "Below 40"]
grade_counts = [
    np.sum(grade_a),
    np.sum(grade_b),
    np.sum(grade_c),
    np.sum(grade_f)
]

plt.figure(figsize=(8, 5))
plt.bar(grade_names, grade_counts)
plt.title("Grade Distribution")
plt.xlabel("Grade Bands")
plt.ylabel("Number of Marks")
plt.show()
```

Student Performance Graph

```python
plt.figure(figsize=(10, 5))
plt.plot(np.arange(1, 51), student_average, marker="o")
plt.title("Student-Wise Average Marks")
plt.xlabel("Student Number")
plt.ylabel("Average Marks")
plt.ylim(0, 100)
plt.show()
```
